# Test calculation for the average cycle length.

```python
# Record the mean of LengthofCycle
dataset_average = cycle_df['LengthofCycle'].mean()

print(f'Dataset average: {dataset_average:.2f} days')
```

```python
# Function to calculate the user average.
def calculate_user_average(cycle_lengths):
    return sum(cycle_lengths) / len(cycle_lengths)


user_cycles = [30, 32, 31]

user_average = calculate_user_average(user_cycles)

print(f'User average: {user_average:.2f}')
```

```python
# Function that uses avergae cycle length from dataset and average user cycle length.
def estimate_cycle_length(dataset_average, user_average, n):
    # User average is weighted because it contains more accuarte data to the user.
    estimated_length = (dataset_average + (user_average * n)) / (1 + n)
    return estimated_length

user_cycles = [30, 32, 31]

user_average = calculate_user_average(user_cycles)
n = len(user_cycles)

estimated_length = estimate_cycle_length(dataset_average, user_average, n)

print(f'Estimated cycle length: {estimated_length:.2f} days')
```

# Predict next period using estimated_length

```python
from datetime import datetime, timedelta

def predict_next_period(last_period_start, estimated_cycle_length):
    cycle_days = round(estimated_cycle_length)

    next_period = last_period_start + timedelta(days=cycle_days)

    return next_period

last_period_start = datetime(2026, 11, 1)

predicted_date = predict_next_period(last_period_start, estimated_length)

print(predicted_date.strftime('%B %d, %Y'))
```
